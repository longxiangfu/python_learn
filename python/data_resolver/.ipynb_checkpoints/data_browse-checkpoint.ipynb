```python
import numpy as np
import pandas as pd
import seaborn as sb
from matplotlib import pyplot as plt
import os
from matplotlib.pylab import mpl
from matplotlib.font_manager import FontProperties

dirpath = 'D:\\workspace\\workspace-python\\python\\data_resolver'
data = pd.read_csv(os.path.join(dirpath, 'athlete_events.csv'))
region = pd.read_csv(os.path.join(dirpath, 'noc_regions.csv'))
```

```python
# 前5行
data.head(5)
```

```python
# 生成描述信息，数据分布的集中趋势，离散度
data.describe()
```

```python
# data的信息
data.info()
```

```python
region.head(5)
```

```python
# 将data和region进行左连接，同mysql的左连接
merged = pd.merge(data, region, on='NOC', how='left')
merged.head(5)
```

```python
# 过滤出金牌数据
goldMedals = merged[merged['Medal'] == 'Gold']
# goldMedals = merged.loc[merged['Medal']=='Gold']
goldMedals.head(100)
```

```python
# 绘制一个金牌获得者年龄分布图，查看数据分布
# 检查列是否有NaN     any()：只要有True的时候返回True     结果中列Age中至少有一个是NaN,列Height Weight也是
goldMedals.isnull().any()
```

```python
# 使用年龄为非NaN的数据   np.isfinite：判断是否为有限值，即不能为无穷大或NaN
# goldMedals = goldMedals[np.isfinite(goldMedals['Age'])]
goldMedals = goldMedals[goldMedals['Age'].notnull]
```

```python
# 绘制
# 画布大小
plt.figure(figsize=(20, 10))

# 绘制图形(普通数量图)    x轴为Age, y轴为对应的数据量
sb.countplot(x= 'Age', data= goldMedals)

# r'' : 转换为原始字符串，避免转义
fontPath = r'D:\workspace\workspace-python\python\data_resolver\毛笔书法字体(启功体)繁启体.TTF'
my_font = FontProperties(fname=fontPath)

# 设置中文字体显示正常 & 不显示负号
mpl.rcParams['axes.unicode_minus'] = False

# 添加图形标题，并应用自定义字体
plt.title('金牌获得者年龄分布', fontproperties=my_font)

# 手动调整布局
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)

# 显示图形
plt.show()
```

```python
# 金牌获得者，年龄大于50，从事的运动项目分布
# 查看符合条件的数据量
# goldMedals[goldMedals['Age'] > 30]  : 过滤出年龄大于50的数据，返回的同样是DataFrame
# goldMedals[goldMedals['Age'] > 30]['ID'] : 获取年龄大于50的数据的ID列，返回的是Series
# count: 返回Series中非NaN的数量
goldMedals[goldMedals['Age'] > 30]['ID'].count()
```

```python
# 绘制
# 过滤数据
older50 = goldMedals[goldMedals['Age'] > 30]

# 画布大小
plt.figure(figsize=(20, 10))


# 绘制图形    x轴为Age, y轴为对应的数据的数量
sb.countplot(x='Sport', data=older50)

# r'' : 转换为原始字符串，避免转义
fontPath = r'D:\workspace\workspace-python\python\data_resolver\毛笔书法字体(启功体)繁启体.TTF'
my_font = FontProperties(fname=fontPath)

# 设置中文字体显示正常 & 不显示负号
mpl.rcParams['axes.unicode_minus'] = False

# 添加图形标题，并应用自定义字体
plt.title('金牌获得者年龄大于30岁的运动项目分布', fontproperties=my_font)

 # 显示图形
plt.show()
```

奥运会随时间的变化

1、男女数量的变化  2、运动员年龄、体重、身高的变化

```python
# 1、男女运动数量随时间的变化（夏季运动会）
MenOverTime = merged[(merged['Sex'] == 'M') & (merged['Season'] == 'Summer')]
WoMenOverTime = merged[(merged['Sex'] == 'F') & (merged['Season'] == 'Summer')]
```

```python
MenOverTime.head()
```

```python
# 按年份性别分组统计
part = MenOverTime.groupby('Year')['Sex'].value_counts()
part_df = pd.DataFrame(part)
part_df.head()
```

```python
part = MenOverTime.groupby('Year')['Sex'].value_counts()
part_df = pd.DataFrame(part)
plt.figure(figsize=(20, 8))
sb.lineplot(x='Year', y='count', data=part_df)
plt.title('男性运动员人数随时间的变化', fontproperties=my_font, fontsize=20)
plt.xlabel('年份', fontproperties=my_font, fontsize=16)
plt.ylabel('男性运动员数量', fontproperties=my_font, fontsize=16)
plt.show()
```

```python
part = WoMenOverTime.groupby('Year')['Sex'].value_counts()
part_df = pd.DataFrame(part)
plt.figure(figsize=(20, 8))
sb.lineplot(x='Year', y='count', data=part_df)
plt.title('女性运动员人数随时间的变化', fontproperties=my_font, fontsize=20)
plt.xlabel('年份', fontproperties=my_font, fontsize=16)
plt.ylabel('女性运动员数量', fontproperties=my_font, fontsize=16)
plt.show()
```

```python
MenOverTime.loc[MenOverTime['Age'] > 80].head(10)
```

```python
# 2、运动员年龄随时间的变化
plt.figure(figsize=(20, 10))
sb.boxplot(x='Year', y='Age', data=MenOverTime)  # 箱型图（盒须图） x为Year y为Age,图中黑色的点为异常值
plt.title('男性运动员年龄随时间的变化', fontproperties=my_font)
plt.show()
```

```python
# 3、运动员体重随时间的变化
plt.figure(figsize=(20, 10))
sb.pointplot(x='Year', y='Weight', data=MenOverTime)# 点状图
plt.title('男性运动员体重随时间的变化', fontproperties=my_font)
plt.show()
```

```python
plt.figure(figsize=(20, 10))
sb.pointplot(x='Year', y='Weight', data=WoMenOverTime)
plt.title('女性运动员体重随时间的变化', fontproperties=my_font)
plt.show()
```
